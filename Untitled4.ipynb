Problem Statement 1: Employee Performance Bonus Eligibility


```python
employees = {
"Ravi": 92,
"Anita": 88,
"Kiran": 92,
"Suresh": 85
}
max_score=max(employees.values())
top_performers=[name for name, score in employees.items() if score == max_score ]
top_performers_str = ", ".join(top_performers)
print(f"Top Performers Eligible for Bonus: {top_performers_str} (Score: {max_score})")
```

Problem Statement 2: Search Query Keyword Analysis


```python

query = "Buy mobile phone buy phone online"
query = query.lower()
query = query.replace(".", "").replace(",", "")
words = query.split()
word_count = {}
for word in words:
    if word in word_count:
        word_count[word] += 1
    else:
        word_count[word] = 1
frequent_words = {}
for word, count in word_count.items():
    if count > 1:
        frequent_words[word] = count
print(frequent_words)
```

Problem Statement 3: Sensor Data Validation


```python
sensor_readings = [3, 4, 7, 8, 10, 12, 5]
valid_readings = []

for hour in range(len(sensor_readings)):
    reading = sensor_readings[hour]
    if reading % 2 == 0:
        valid_readings.append((hour, reading))

print("Valid Sensor Readings (Hour, Value):")
print(valid_readings)
```

Problem Statement 4: Email Domain Usage Analysis


```python
emails = [
    "ravi@gmail.com",
    "anita@yahoo.com",
    "kiran@gmail.com",
    "suresh@gmail.com",
    "meena@yahoo.com"
]


domain_count = {}
for email in emails:
    domain = email.split("@")[1]
    if domain in domain_count:
        domain_count[domain] += 1
    else:
        domain_count[domain] = 1


total_emails = len(emails)


for domain, count in domain_count.items():
    percentage = (count / total_emails) * 100
    print(f"{domain}: {int(percentage)}%")
```

Problem Statement 5: Sales Spike Detection


```python
sales = [1200, 1500, 900, 2200, 1400, 3000]


average_sales = sum(sales) / len(sales)

threshold = average_sales * 1.3

for day in range(len(sales)):
    if sales[day] > threshold:
        print(f"Day {day + 1}: {sales[day]}")
```

Problem Statement 6: Duplicate User ID Detection


```python
user_ids = ["user1", "user2", "user1", "user3", "user1", "user3"]
user_count = {}
for user in user_ids:
    if user in user_count:
        user_count[user] += 1
    else:
        user_count[user] = 1


for user, count in user_count.items():
    if count > 1:
        print(f"{user} → {count} times")
```
